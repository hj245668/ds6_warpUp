일별 수익율 구하기
- 수익률 : 돈을 얼마나 벌거나 잃었는지 나타냄
- 일별 수익율 : 하루동안의 수익율
> / 양수 수익률 : 자산이 상승했음을 의미함
> / 음수 수익률 : 자산이 하락했음을 의미함
> / 0 수익률 : 자산의 가격이 변하지 않았음
- 일별 수익률 (%) = 100 * ((오늘가격 - 어제가격) / 어제 가격 )
             


```python
# !pip install --upgrade finance-datareader


# # 필요한 라이브러리를 가져오기
# import matplotlib.pyplot as plt  # Matplotlib를 가져옴
# import pandas as pd  # Pandas를 가져옴
# import FinanceDataReader as fdr  # FinanceDataReader를 가져옴
```

```python
# 첫 번째 인자로 종목 코드 '005930'을 전달하고,
# 두 번째 인자로 시작일을 '2015-01-01'로 지정합니다.

df = fdr.DataReader('005930', '2015-01-01')
# (종목, 시작일)
# DataFrame의 head() 메서드를 사용하여 처음 5개 행을 표시합니다.
df.head(5)
```

```python
# 주식 데이터프레임(df)에서 'Close' 열의 데이터를 추출하고,
# 이를 사용하여 일별 수익률을 계산하는 코드입니다.

# .pct_change(): 현재값과 이전값의 백분율, 일별 수익률을 계산하는 과정
# 현재 값과 이전 값 사이의 백분율 변화를 나타냅니다
# ( 계산된 백분율 변화율을 백분율로 표시하기 위해 100을 곱합니다)

df = (df['Close'].pct_change()) * 100
```

```python
df.describe()
```

```python
plt.figure(figsize=(12,6))

# plt.plot(r.index, r, label='Daily Returns', color='b', linestyle='-'):
# 'df' 데이터를 이용하여 그래프를 그립니다.
# df.index: 'r' 데이터의 인덱스(날짜)를 x 축으로 사용합니다.
# df: 일별 수익률 데이터를 y 축으로 사용합니다.
# color='b': 그래프의 색상을 파란색(blue)으로 설정합니다.
# linestyle='-': 그래프의 선 스타일을 실선으로 설정합니다.

plt.plot(df.index, df, color = 'b', linestyle='-')
# 그래프 제목
plt.title("daily returns of samsung electronics(005930) stock")

# x 축 레이블
plt.xlabel('date')

# y 축 레이블
plt.ylabel('daily returns %')

# 그래프 그리드 시각확
plt.grid(True)

#그래프 표시
plt.show()
```

```python
df1 = df.loc['2022-01-01': '2022-12-31']
df2=df1.copy()
df2['ret'] = (df2['Close'].pct_change())*100
df3 = df2['ret']
df3
```

```python
r = df3.dropna()
len(r)
```

```python
# 일별 수익율이 양인것의 갯수 : positive
```

```python
# 245개의 일별 수익율에서 양의 수익율을 positive라고 하고, 양의 수익율 횟수를 누적함

positive = 0
n = len(r)

for i in range(n):
  if r[i] > 0 :
    positive += 1
positive
```

```python
negative = n - positive
```

```python
negative
```

```python
positive_rate = positive / n
```

```python
positive_rate
```

특정 구간에 주가가 오를 수 있는 구간 확인

- 조합공식 : 가능한 경우의 수를 계산
- 이항분포 : 조합의 각 경우가 일어날 확률을 함께 곱해 특정 성공 횟수의 확률을 구하는 분포
- 활용 : 주식 상승 확률 모델이나 거래일 구간별 성공률 예측
- 이론 참고 : https://math100.tistory.com/19?utm_source=chatgpt.com
- stock 분석 참고 : https://chartschool.stockcharts.com/table-of-contents/overview/technical-analysis?utm_source=chatgpt.com

```python
import math
a = math.factorial(5)
b = math.factorial(3)*math.factorial(2)
number = a/b
# 5일간의 변동을 이용해 3일 예

c = ((positive_rate)**3)*((1-positive_rate)**2)
# 성공확률 positive_rate 이 2번 발생하고,
# 실패확률 ( 1-positive_rate)이 2번 발생하는 확률을 계산합니다.

prob = number * c
prob
```
